## 3. Revenue by City

```python
city_index = 2
revenue_index = 6

city_revenue = {}

for row in sales_data[1:]:
    city = row[city_index]
    revenue = float(row[revenue_index])

    if city in city_revenue:
        city_revenue[city] += revenue 
    else:
        city_revenue[city] = revenue

print("Total Revenue by City:")
for city, total in city_revenue.items() :
     print(f"- {city}: {total:,.2f} SEK")
```

## 4. Revenue by Category

```python
category_index = 3
revenue_index = 6

category_revenue = {}

for row in sales_data[1:]:
    category = row[category_index]
    revenue = float(row[revenue_index])

    if category in category_revenue:
        category_revenue[category] += revenue
    else:
        category_revenue[category] = revenue

print("Total Revenue by Category:") 
for category, total in category_revenue.items():
    print(f"- {category}: {total:,.2f} SEK")      
```

## 5. Total Overall Revenue

```python
revenue_index = 6
total_overall_revenue = 0

for row in sales_data[1:]:
    total_overall_revenue += float(row[revenue_index])

print(f"Total Overall Revenue: {total_overall_revenue:,.2f} SEK")
```
