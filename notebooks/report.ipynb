## 3. Revenue by City

```python
category_index = 3
revenue_index = 6

category_revenue = {}

for row in sales_data[1:]:
    category = row[category_index]
    revenue = float(row[revenue_index])

    if category in category_revenue:
        category_revenue[category] += revenue
    else:
        category_revenue[category] = revenue

print("Total Revenue by Category:") 
for category, total in category_revenue.items():
    print(f"- {category}: {total:,.2f} SEK")      
```

Analys av försäljningsmönster över tid.

```python
import matplotlib.pyplot as plt
import pandas as pd

df_time = pd.read_csv('../data/ecommerce_sales.csv')

df_time['date'] = pd.to_datetime(df_time['date'])

monthly_sales = df_time.groupby(df_time['date'].dt.to_period('M'))['revenue'].sum()

plt.figure(figsize=(9, 4))
monthly_sales.plot(kind='line', marker='o', color='#2b5c8f', linewidth=2)
plt.title('Monthly Sales Trend (När säljer vi?)')
plt.xlabel('Month')
plt.ylabel('Total Revenue (SEK)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

print("Monthly Revenue Breakdown:")
for month, rev in monthly_sales.items():
    print(f"- {month}: {rev:,.2f} SEK")

```

**Slutsats:**
Försäljningen visar en tydlig variation över perioden med en start 1 januari (ca 664,083 SEK) och en ny uppgång 1 maj (ca 632,165 SEK). Den lägsta försäljningsvolymen under perioden noterades 1 mare.
